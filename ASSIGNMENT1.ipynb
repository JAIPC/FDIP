```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("car_fdip.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(6, 6))
plt.title("Original Image")
plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
gray_img = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
plt.figure(figsize=(6, 6))
plt.title("Grayscale Image")
plt.imshow(gray_img, cmap="gray")
plt.axis("off")
plt.show()
```

```python
bright_plus = cv2.convertScaleAbs(gray_img, alpha=1.0, beta=40)
bright_minus = cv2.convertScaleAbs(gray_img, alpha=1.0, beta=-40)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.title("Brightness Increased")
plt.imshow(bright_plus, cmap="gray")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.title("Brightness Decreased")
plt.imshow(bright_minus, cmap="gray")
plt.axis("off")
plt.show()
```

```python
negative_img = cv2.bitwise_not(gray_img)
plt.figure(figsize=(6, 6))
plt.title("Negative Image")
plt.imshow(negative_img, cmap="gray")
plt.axis("off")
plt.show()
```

```python
_, binary_img = cv2.threshold(gray_img, 128, 255, cv2.THRESH_BINARY)
plt.figure(figsize=(6, 6))
plt.title("Binary Thresholding")
plt.imshow(binary_img, cmap="gray")
plt.axis("off")
plt.show()
```

```python
log_transformed = np.log1p(gray_img.astype(np.float32))
log_transformed = cv2.normalize(log_transformed, None, 0, 255, cv2.NORM_MINMAX)
log_transformed = np.uint8(log_transformed)

plt.figure(figsize=(6, 6))
plt.title("Log Transformation")
plt.imshow(log_transformed, cmap="gray")
plt.axis("off")
plt.show()
```

```python
gamma_value = 0.5
gamma_corrected = np.power(gray_img / 255.0, gamma_value)
gamma_corrected = np.uint8(gamma_corrected * 255)

plt.figure(figsize=(6, 6))
plt.title(f"Gamma Correction (γ={gamma_value})")
plt.imshow(gamma_corrected, cmap="gray")
plt.axis("off")
plt.show()
```
