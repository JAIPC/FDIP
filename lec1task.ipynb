```python
import numpy as np
import matplotlib.pyplot as plt 
from skimage import img_as_uint
from skimage.io import imshow, imread
from skimage.color import rgb2hsv
from skimage.color import rgb2gray
```

```python
array_spectrum = np.array([np.arange(0,255,17),np.arange(255,0,-17),np.arange(0,255,17),np.arange(255,0,-17)])
fig,ax = plt.subplots(1, 2,figsize=(12,4))
ax[0].imshow(array_spectrum, cmap = 'gray')
ax[0].set_title('Arange Generation')
ax[1].imshow(array_spectrum.T, cmap = 'gray')
ax[1].set_title('Transpose Generation');
```

```python
car=imread(r"C:\Users\ML\Downloads\download.png")
imshow(car);
```

```python
fig,ax = plt.subplots(1, 3,figsize=(6,4),sharey= False)

ax[0].imshow(car[:, 0:1000])
ax[0].set_title('First Split')

ax[1].imshow(car[:, 1000:2000])
ax[1].set_title('Second Split')

ax[2].imshow(car[:, 2000:3000])
ax[2].set_title('Third Split');
```

```python
imshow(car[1500:2500, 1500:3000]);
```

```python
fig, ax = plt.subplots(1, 3, figsize=(12,4), sharey = True)
ax[0].imshow(car[:,:,0], cmap='Reds')
ax[0].set_title('Red')
ax[1].imshow(car[:,:,1], cmap='Greens')
ax[1].set_title('Green')
ax[2].imshow(car[:,:,2], cmap='Blues')
ax[2].set_title('Blue');
```

```python
car_hsv = rgb2hsv(car)
fig, ax = plt.subplots(1, 3, figsize=(12,4), sharey = True)
ax[0].imshow(car_hsv[:,:,0], cmap='hsv')
ax[0].set_title('Hue')
ax[1].imshow(car_hsv[:,:,1], cmap='gray')
ax[1].set_title('Saturation')
ax[2].imshow(car_hsv[:,:,2], cmap='gray')
ax[2].set_title('Value');
```

```python
car_gray = rgb2gray(car)
fig, ax =plt.subplots(1, 5, figsize=(17,6), sharey = True)

ax[0].imshow(car_gray, cmap = 'gray')
ax[0].set_title('GrayscaleOriginal')

ax[1].imshow(img_as_uint(car_gray >0.25),cmap = 'gray')
ax[1].set_title('Greater than0.25')

ax[2].imshow(img_as_uint(car_gray >0.50),cmap = 'gray')
ax[2].set_title('Greater than0.50');

ax[3].imshow(img_as_uint(car_gray >0.75),cmap = 'gray')
ax[3].set_title('Greater than0.75')

ax[4].imshow(img_as_uint(car_gray >np.mean(car_gray)),cmap = 'gray')
ax[4].set_title('Greater than Mean');
```
